```python
from slither.slither import Slither
import os
from dotenv import load_dotenv
import sys
sys.path.append("../")

load_dotenv()
from src.detectors import AC
from src.response.AC_response import build_access_control_response
```

```python
def run_AC_module(chain: str, address: str) -> dict:

    prefix=''
    if chain=="arbitrum": prefix = 'arbi:'
    if chain=="base": prefix = 'base:'
    if chain=='optimism': prefix = 'optim:'

    try:
        sl = Slither(prefix+address, etherscan_api_key=os.getenv("ETHERSCAN_API_KEY"))
    except Exception as e :
        print(e)
        raise(e)

    AC_results = AC.run(sl)

    response_AC = build_access_control_response(chain, address, AC_results )
  
    return AC_results, response_AC
```

```python
# run on local file : 
! solc-select use 0.8.20
result, response = run_AC_module(chain='local', address='../test-contracts/VulnerableMerkleTest.sol')

print(result)

response
```
